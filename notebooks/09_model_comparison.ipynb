# Final Model Comparison

Run this LAST, after running whichever experiment notebooks you want included. Reads every `results/E*.csv`, combines them, and produces the comparison table + charts.

```
import pandas as pd
import glob
import matplotlib.pyplot as plt
import numpy as np

result_files = sorted(glob.glob("../results/E*.csv"))
print(f"Found {len(result_files)} result files:")
for f in result_files:
    print(" ", f)

results_df = pd.concat([pd.read_csv(f) for f in result_files], ignore_index=True)
results_df = results_df.set_index("run_name")
results_df
```

```
results_df.to_csv("../results/comparison_all.csv")
print("Saved combined comparison to results/comparison_all.csv")
```

## Accuracy comparison

```
fig, ax = plt.subplots(figsize=(10, 5))
results_df["accuracy"].sort_values().plot(kind="barh", ax=ax, color="steelblue")
ax.set_xlabel("Test Accuracy")
ax.set_title("Model Comparison: Accuracy")
for i, v in enumerate(results_df["accuracy"].sort_values()):
    ax.text(v + 0.005, i, f"{v:.2%}", va="center", fontsize=9)
plt.tight_layout()
plt.savefig("../results/accuracy_comparison.png", dpi=150)
plt.show()
```

## All metrics, grouped

```
metrics_to_plot = ["accuracy", "precision", "recall", "f1", "roc_auc"]
models = results_df.index.tolist()
x = np.arange(len(models))
width = 0.15

fig, ax = plt.subplots(figsize=(14, 6))
for i, metric in enumerate(metrics_to_plot):
    if metric in results_df.columns:
        ax.bar(x + i * width, results_df[metric].values, width, label=metric)

ax.set_xticks(x + width * (len(metrics_to_plot) - 1) / 2)
ax.set_xticklabels(models, rotation=30, ha="right")
ax.set_ylabel("Score")
ax.set_title("Model Comparison Across Metrics")
ax.set_ylim(0, 1.05)
ax.legend(loc="lower right")
ax.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.savefig("../results/grouped_comparison.png", dpi=150)
plt.show()
```

## Training time comparison

```
fig, ax = plt.subplots(figsize=(10, 5))
results_df["train_time_sec"].sort_values().plot(kind="barh", ax=ax, color="indianred")
ax.set_xlabel("Training time (seconds)")
ax.set_title("Model Comparison: Training Time")
plt.tight_layout()
plt.savefig("../results/train_time_comparison.png", dpi=150)
plt.show()
```

## Reasoning notes

- **Best model:** _which run had the highest accuracy/F1, and by how much vs. baseline?_
- **Biggest single win:** _which one enhancement gave the largest jump?_
- **Any regressions:** _did any enhancement make things worse? Why?_
- **Cost vs. benefit:** _which run took longest, and was the accuracy gain worth it?_
